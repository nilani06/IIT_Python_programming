#### For loop

```python
# for variable in range(start, stop, step):
for i in range(5): # 5 means the stop value, by default start is 0 and step is 1 (0, 5, 1) 
    print(i)
```

```python
for i in range(10, 21, 2):
    print(i)
```

#### While loop

```python
i = 0
while i < 5:
    print(i)
    i += 1
```

```python
for i in range(2, 11, 2): # even numbers from 2 ro 10
    print(i)
```

```python
for i in range(5):
    print('Hello!')
```

```python
for i in range(1, 36, 2): # odd numbers upto 35
    print(i)
```

```python
fruit_List = ['apple', 'banana', 'cherry']
for fruit in fruit_List:
    print(fruit)
```

```python
student = {"name": "Nilani", "age": 19, "grade": "A"}

for key, value in student.items():
    print(f"{key}: {value}")
```
