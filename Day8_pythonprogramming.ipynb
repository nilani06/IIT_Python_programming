### Exception Handling

```python
# How to debug
try:
    print(a)
except:
    print("An error occurred")
# print ("An error occurred") instead of crashing the program
```

```python
try:
    x = 10/0
except:
    print("You cannot divide a number by zero")
```

```python
# else block - runs when no error occurs
# finally block - always runs
# try + except              (Allowed)
# try + finally             (Allowed)
# try + except + finally    (Allowed)
# try + eccept + else       (Allowed)
# try + else                (Not Allowed)
# try alone                 (Not Allowed)
```

```python
try:
    num=int(input("Enter a number: "))
    print(num)
except:
    print("Incorrect integer format")

# if the erroe didn't occur
else:
    print("Success! You entered: ",num)
```

```python
try:
    num=int(input("Enter a number: "))
    print(num)
except:
    print("Incorrect integer format")

# if error occurs or not, finally) will still run
finally:
    print("Try again")
```

### Homework

#### Write a Python program that uses a try-except block to safely read marks from the user.
#### Your program must:
#### - Ask the user to enter the marks.
#### - Use a try block to convert the input into an integer.
#### - Use an except block to catch errors if the user enters invalid data (non-numeric input). "Invalid input! Please enter a number."
#### - If the number is valid but outside the range (0-100), print an error message: "Marks must be between 0 and 100."
#### - If the number is valid and within range, print "Valid marks entered!".

```python
try:
    marks=int(input("Enter Marks: "))
    if 0 <= marks <= 100:
        print("Valid marks entered!")
    else:
        print("Marks must be between 0 and 100.")
except ValueError:
    print("Invalid input! Please enter a number.")
```
