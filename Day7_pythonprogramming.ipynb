#### sets and Dictionaries

```python
my_set={1,3,2,4,4,5,2,1}
print(type(my_set))
print(my_set) # Sets do not print duplicates, and it prints in order
```

```python
my_set={1,2,3}
my_set.add(4) # add a single element
my_set.update([5,6]) # add multiple elements
my_set.remove(2) # remove element (error if not found)
print(my_set)
my_set.discard(10) # remove element (no error if not found) 
my_set.pop() # remove a random element
print(my_set)
my_set.clear() # remove all element
print(my_set)
```

```python
my_set3={101,102,103,104,105,106,107,108}
my_set3.add(109)
print(my_set3)
my_set3.remove(101)
print(my_set3)
my_set3.discard(200)
print(my_set3)
my_set3.pop()
print(my_set3)
my_set3.clear()
print(my_set3)
```

```python
A={1,2,3}
B={3,4,5}
print(A.union(B))
print(B.union(A))
#or
print(A|B)
print(B|A)

print(A.intersection(B))
print(B.intersection(A))
#or
print(A&B)
print(B&A)

print(A.difference(B))
print(B.difference(A))
#or
print(A-B)
print(B-A)
```
