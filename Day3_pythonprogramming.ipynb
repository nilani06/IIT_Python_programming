```python
#if-else
number=int(input('Enter a number'))
if number>0:
    print("Positive")
elif number<0:
    print("Negative")
else:
    print("Zero")
```

```python
#Nested if
number=int(input('Enter a number'))
if number>0:
    print("Positive")
else:
    if number!=0:
        print("Negative")
    else:
        print("Zero")
```

```python
number=int(input('Enter a number'))
if number%2==1:
    print('odd')
else:
    print('even')
```

```python
student_name=input('Enter Student name')
student_marks=int(input('Enter your marks: '))
if student_marks>=75:
    print(f'Amazing {student_name} you have an A pass')
elif student_marks>=65:
    print(f'Good {student_name} you have a B pass')
elif student_marks>=50:
    print(f'Good {student_name} you have a C pass') 
else:
    print(f'Sorry {student_name} you have a F Pass, try again')
```

```python
student_marks=int(input('Enter your mark out of 100: '))
while True:
    if student_marks<=0 or student_marks>=100:
        student_marks=int(input('Invalid, enter again: '))
    else:
        break
if student_marks>=90:
    print('Grade A')
elif student_marks>=80:
    print('Grade B')
elif student_marks>=70:
    print('Grade C')
elif student_marks>=60:
    print('Grade D')
else:
    print('Fail')
```
