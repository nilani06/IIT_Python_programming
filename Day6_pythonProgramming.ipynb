```python
fruits=["apple","banana","mango"]
fruits[1]="orange"
print(fruits)
fruits.append("banana")
print(fruits)
```

```python
fruits_2=["apple","banana","mango","orange","cherry"]
fruits_2.remove("orange") #search and remove
print(fruits_2)
fruits_2.pop() #removes last 
print(fruits_2)
fruits_2.pop(1) #removes with index
print(fruits_2)
fruits_2.clear() #removes all
print(fruits_2)
```

```python
my_list=[20,"john",34,False,"Anne"]
print(my_list[0]) #first element
print(my_list[:3]) #first 3
print(my_list[::-1]) #reverst
print(my_list[:]) #ful copy
print(my_list[2:5]) #slice from 2 to 5
print(my_list[2:]) #slice from 2 to end
print(my_list[2:5:2]) #slice from 2 to 5 with step
```
